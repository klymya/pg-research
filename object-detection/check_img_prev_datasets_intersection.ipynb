```python
import json
import os

import numpy as np
```

```python
with open("annotation.json", 'rt') as f:
    data = json.load(f)
img_names = set([i.replace(".bmp", "") for i in data.keys()])
len(img_names)
```

```python
def read_dataset(path):
    with open(os.path.join(path, "train_image.txt"), 'rt') as f:
        lines = f.readlines()
    train_lines = [os.path.basename(i).replace(".jpg\n", "").replace(".bmp\n", "").replace(".jpeg\n", "") for i in lines]

    with open(os.path.join(path, "val_image.txt"), 'rt') as f:
        lines = f.readlines()
    val_lines = [os.path.basename(i).replace(".jpg\n", "").replace(".bmp\n", "").replace(".jpeg\n", "") for i in lines]
    
    return set(train_lines), set(val_lines)
```

```python
train_1238, val_1238 = read_dataset('../../deep-learning-training/dataset_list/Finetune_set_1238_list/')
```

```python
train_550, val_550 = read_dataset('../../deep-learning-training/dataset_list/Finetune_set_550_list/')
```

```python
len(train_1238), len(val_1238)
```

```python
len(train_550), len(val_550)
```

```python
len(set(train_1238).intersection(val_1238)), len(set(train_550).intersection(val_550))
```

```python
val_1238 = set(val_1238).difference(train_1238)
```

```python
len(set(val_1238))
```

```python
len(train_1238.intersection(img_names)), len(val_1238.intersection(img_names))
```

```python
test_set = val_1238.intersection(img_names)
train_set = set(img_names).difference(test_set)
len(train_set)
```

```python
idxs = np.arange(len(train_set))
np.random.seed(42)
np.random.shuffle(idxs)
n = 95
train_idxs = idxs[:-n]
val_idxs = idxs[-n:]
len(train_idxs), len(val_idxs)
```

```python
with open('baseline_train.txt', 'wt') as f:
    for i in np.array(list(train_set))[train_idxs]:
        f.write(f'{i}\n')
        
with open('baseline_val.txt', 'wt') as f:
    for i in np.array(list(train_set))[val_idxs]:
        f.write(f'{i}\n')
        
with open('baseline_test.txt', 'wt') as f:
    for i in test_set:
        f.write(f'{i}\n')
```
