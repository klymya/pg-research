```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
with open("tesnsor_output.txt", "rt") as f:
    data = f.readlines()[0]

data = [int(i) for i in data.split(" ")]
len(data)

data = np.array(data)
```

```python
plt.imshow(data.reshape(240, 320, -1), cmap='gray')
```

```python
plt.imshow(data[::3].reshape(240, 320), cmap='gray')
```
