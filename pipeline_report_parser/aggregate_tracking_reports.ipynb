```python
BASE_PATH = "/mnt/pg-fs02/benchmarks-analysis"
BASE_SUFFIX = "TrackingAnalysis/TrackingAnalysisResults.csv"

DATASETS = [
    "PG1_AlgoDev_DLtest02_RV9",
    "PG1_AlgoDev_DLtest_RV59",
    "PG1_AlgoDev_VLC_RV5_V2",
    "10fps",
]

# Requre setup
RUN_IDS = [
    "20250408_020344_5_3_0+gitv6_2_0-rc_1-1-g9ec5a865_PG1_AlgoDev_DLtest02_RV9",
    "20250408_032251_5_3_0+gitv6_2_0-rc_1-1-g9ec5a865_PG1_AlgoDev_DLtest_RV59",
    "20250408_051713_5_3_0+gitv6_2_0-rc_1-1-g9ec5a865_PG1_AlgoDev_VLC_RV5_V2",
    "20250408_065922_5_3_0+gitv6_2_0-rc_1-1-g9ec5a865_10fps",
]

file_paths = [
    os.path.join(BASE_PATH, dataset, run_id, BASE_SUFFIX) for dataset, run_id in zip(DATASETS, RUN_IDS)
]

file_paths = [
    "/home/klymy/projects/researches/pg1_yolo_evaluator/10fps/TrackingAnalysis/TrackingAnalysisResults.csv",
    "/home/klymy/projects/researches/pg1_yolo_evaluator/PG1_AlgoDev_DLtest_RV59/TrackingAnalysis/TrackingAnalysisResults.csv",
    "/home/klymy/projects/researches/pg1_yolo_evaluator/PG1_AlgoDev_DLtest02_RV9/TrackingAnalysis/TrackingAnalysisResults.csv",
    "/home/klymy/projects/researches/pg1_yolo_evaluator/PG1_AlgoDev_VLC_RV5_V2/TrackingAnalysis/TrackingAnalysisResults.csv",
]
```

```python
dfs = []
for file_pth in file_paths:
    if os.path.exists(file_pth):
        dfs.append(pd.read_csv(file_pth))
    else:
        print(f"File not found: {file_pth}")

df = pd.concat(dfs, ignore_index=True)

df.shape
```

```python
print("Number of rows:", df.shape[0])
print("=== Mean ===")
print('F1 score      :', df["F1 score"].mean().round(2))
print('Precision     :', df["Precision"].mean().round(2))
print('Recall        :', df["Recall"].mean().round(2))
print("=== Median ===")
print('F1 score      :', df["F1 score"].median().round(2))
print('Precision     :', df["Precision"].median().round(2))
print('Recall        :', df["Recall"].median().round(2))
```
