````python
import functools
import os
import sys

from ultralytics import YOLO
import pandas as pd

sys.path.append("..")
import ultralytics_validator_wrapper

pd.options.display.float_format = '{:.3f}'.format

%matplotlib inline
%load_ext autoreload
%autoreload 2
````

````python
BASE_WEIGHTS_PTH = "../results/filtered_v3_margin_15/weights/best.pt"
EXP_WEIGHTS_PTH = "../results/filtered_v3_margin_15_reduced_quality/weights/best.pt"

CONF_TH = 0.35
IOU_TH = 0.5

args = dict(
    task="pose",
    device=[0],
    batch=128,
    verbose=False,
    conf=CONF_TH,
    iou=IOU_TH,
    imgsz=320,
    model="val",
    rect=False,
    single_cls=False,
)

model = YOLO(BASE_WEIGHTS_PTH)
model_exp = YOLO(EXP_WEIGHTS_PTH)
````

````python
BASE_PTH = "/home/kyamkovyi/projects/researches/people_detection_pg2/quality_reduce_comparison/"

DATASETS = [
    "cam4_poc_part1",
    "cam4_poc_part1_15_blurred",
    "cam4_poc_part1_15_grayscale",
    "cam4_poc_part1_15_blurred_grayscale",
]


for dataset in DATASETS:
    print("=" * 80)
    print(f"DATASET: {dataset}")
    print("=" * 80)
    VAL_PTH = os.path.join(BASE_PTH, dataset, "project.yaml")
    args["data"] = VAL_PTH

    print("BASE MODEL")
    ultralytics_validator_wrapper.validate_model(
        args,
        model,
        functools.partial(ultralytics_validator_wrapper.predict_ultralytics, model, args),
        False
    )

    # print("EXPERIMENT MODEL")
    # ultralytics_validator_wrapper.validate_model(
    #     args,
    #     model_exp,
    #     functools.partial(ultralytics_validator_wrapper.predict_ultralytics, model_exp, args),
    #     False
    # )
    print()
````
