```python
utils_cam4_revb.cp_dataset(
    base_pth=BASE_PTH,
    src=BASE_NAME,
    dst=RESULTING_NAME,
)
```

```python
# cp train, val

subfolders = [
    "images/train",
    "images/val",
    "labels/train",
    "labels/val",
]

for subfolder in subfolders:
    src = os.path.join(BASE_PTH, UPD_NAME, subfolder)
    dst = os.path.join(BASE_PTH, RESULTING_NAME, subfolder)

    for f in glob.glob(os.path.join(src, "*")):
        shutil.copy(f, dst)
```

```python
# cp holdout

shutil.copytree(
    os.path.join(BASE_PTH, RESULTING_NAME, "labels", "holdout"),
    os.path.join(BASE_PTH, RESULTING_NAME, "labels", "holdout_v2"),
    dirs_exist_ok=True,
)
shutil.copytree(
    os.path.join(BASE_PTH, RESULTING_NAME, "images", "holdout"),
    os.path.join(BASE_PTH, RESULTING_NAME, "images", "holdout_v2"),
    dirs_exist_ok=True,
)

for subfolder in ["images", "labels"]:
    src = os.path.join(BASE_PTH, UPD_NAME, subfolder, "holdout")
    dst = os.path.join(BASE_PTH, RESULTING_NAME, subfolder, "holdout_v2")

    for f in glob.glob(os.path.join(src, "*")):
        shutil.copy(f, dst)

utils_cam4_revb.create_project_file(
    os.path.join(os.path.join(BASE_PTH, RESULTING_NAME), "project_holdout_v2.yaml"),
    os.path.join(BASE_PTH, RESULTING_NAME),
    {0: "person"},
    "images/holdout_v2")
```
