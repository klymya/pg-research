```python
import functools
import os
import sys

from ultralytics import YOLO
import pandas as pd

sys.path.append("..")
import ultralytics_validator_wrapper

pd.options.display.float_format = '{:.3f}'.format

%matplotlib inline
%load_ext autoreload
%autoreload 2
```

```python
# BASE_WEIGHTS_PTH = "../results/filtered_v3_margin_15/weights/best.pt"
# EXP_WEIGHTS_PTH = "../results/filtered_v3_margin_15_reduced_quality/weights/best.pt"
BASE_WEIGHTS_PTH = "../results/filtered_v3_margin_15_yolo11n/weights/best.pt"
EXP_WEIGHTS_PTH = "../results/filtered_v3_margin_15_reduced_quality_yolo11n/weights/best.pt"

CONF_TH = 0.25
IOU_TH = 0.55

args = dict(
    task="pose",
    device=[0],
    batch=128,
    verbose=False,
    conf=CONF_TH,
    iou=IOU_TH,
    imgsz=320,
    model="val",
    rect=False,
    single_cls=False,
)

model = YOLO(BASE_WEIGHTS_PTH)
model_exp = YOLO(EXP_WEIGHTS_PTH)
```

```python
pth = "../quality_reduce_comparison/"
BASE_PTH = os.path.abspath(pth)

DATASETS = [
    "cam4_poc_15",
    "cam4_poc_15_blurred_grayscale",
    # "cam4_poc_15_nn",
]
YAML_NAME = "project_pg.yaml"
```

```python
for dataset in DATASETS:
    print("=" * 80)
    print(f"DATASET: {dataset}")
    print("=" * 80)
    VAL_PTH = os.path.join(BASE_PTH, dataset, YAML_NAME)
    args["data"] = VAL_PTH

    print("BASE MODEL")
    ultralytics_validator_wrapper.validate_model(
        args,
        model,
        functools.partial(ultralytics_validator_wrapper.predict_ultralytics, model, args),
        False
    )

    # print("EXPERIMENT MODEL")
    # ultralytics_validator_wrapper.validate_model(
    #     args,
    #     model_exp,
    #     functools.partial(ultralytics_validator_wrapper.predict_ultralytics, model_exp, args),
    #     False
    # )
    print()
```
