```python
import functools
import os
import sys

from ultralytics import YOLO
import pandas as pd

sys.path.append("..")
import ultralytics_validator_wrapper

pd.options.display.float_format = '{:.3f}'.format

%matplotlib inline
%load_ext autoreload
%autoreload 2
```

```python
BASE_PTH = os.path.abspath(".")

YAML_NAME = "project_pg.yaml"

DATASETS = [
    "cam4_poc_15",
    "cam4_poc_15_blurred",
    "cam4_poc_15_grayscale",
    "cam4_poc_15_blurred_grayscale",
]

BASE_WEIGHTS_PTH = "../results/filtered_v3_margin_15_uniform_noise/weights/best.pt"

CONF_TH = 0.25
IOU_TH = 0.55

args = dict(
    task="pose",
    device=[1],
    batch=128,
    verbose=False,
    conf=CONF_TH,
    iou=IOU_TH,
    imgsz=320,
    model="val",
    rect=False,
    single_cls=False,
)

model = YOLO(BASE_WEIGHTS_PTH)
```

```python
for dataset in DATASETS:
    print("=" * 80)
    print(f"DATASET: {dataset}")
    print("=" * 80)
    VAL_PTH = os.path.join(BASE_PTH, dataset, YAML_NAME)
    args["data"] = VAL_PTH

    ultralytics_validator_wrapper.validate_model(
        args,
        model,
        functools.partial(ultralytics_validator_wrapper.predict_ultralytics, model, args),
        False
    )
    print()
```
