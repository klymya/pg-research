```python
import os
import shutil

from ultralytics import YOLO
import matplotlib.pyplot as plt
import cv2
import tqdm

%matplotlib inline
```

```python
model = YOLO("results/merged_furniture_no_occupied_chairs/weights/best.pt")
```

```python
iou = 0.5
conf = 0.25
```

```python
dataset_pth = "datasets/merged_furniture/images/val"
results_pth = "viz_debug_no_chairs"

shutil.rmtree(f"{results_pth}", ignore_errors=True)
os.makedirs(results_pth)

for img_pth in tqdm.tqdm(os.listdir(dataset_pth)):
    img_pth = os.path.join(dataset_pth, img_pth)

    plt.figure(figsize=(20, 7))
    plt.title(os.path.basename(img_pth), fontsize=15)

    results = model(img_pth, verbose=False, iou=iou, conf=conf)
    # Visualize the detection
    img = results[0].plot()  # This plots the detections on the image
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))

    plt.savefig(os.path.join(results_pth, os.path.basename(img_pth).replace(".bmp", ".jpg")))
    plt.close()
```

```python
base_pth = "results"
model_names = [
    "merged_furniture",
    "merged_furniture_no_occupied",
    "merged_furniture_no_occupied_chairs"
]

models = []
for model_name in model_names:
    models.append(YOLO(f"{base_pth}/{model_name}/weights/best.pt"))


dataset_pth = "datasets/merged_furniture/images/val"
results_pth = "viz_debug_all"

shutil.rmtree(f"{results_pth}", ignore_errors=True)
os.makedirs(results_pth)

for img_pth in tqdm.tqdm(os.listdir(dataset_pth)):
    img_pth = os.path.join(dataset_pth, img_pth)

    plt.figure(figsize=(15, 5))
    # plt.suptitle(os.path.basename(img_pth), fontsize=15)
    plt.tight_layout()
    plt.subplots_adjust(wspace=0.1, hspace=0.1)
    for i, model in enumerate(models):
        results = model(img_pth, verbose=False, iou=iou, conf=conf, task="pose")
        # Visualize the detection
        img = results[0].plot()
        plt.subplot(1, len(models), i + 1)
        plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
        plt.title(model_names[i])

    plt.savefig(os.path.join(results_pth, os.path.basename(img_pth).replace(".bmp", ".jpg")))
    plt.close()
    # break
```
