```python
tmp_img = cv2.imread(img_paths[111])
#tmp_img = cv2.imread(img_paths[9501])
plt.imshow(cv2.cvtColor(tmp_img, cv2.COLOR_BGR2RGB))
```

```python
def blur(img, kernel_size=(15, 15)):
    return cv2.GaussianBlur(img, kernel_size, 0)

def adjast_contrast_brightness(img, alpha=1, beta=0):
    return cv2.convertScaleAbs(img, alpha=alpha, beta=beta)

def adjast_gamma(img, gamma=1):
    invGamma = 1.0 / gamma
    table = np.array([((i / 255.0) ** invGamma) * 255 for i in np.arange(0, 256)]).astype("uint8")
    return cv2.LUT(img, table)
```

```python
plt.imshow(cv2.cvtColor(blur(tmp_img.copy()), cv2.COLOR_BGR2RGB))
```

```python
blurred_img = blur(tmp_img.copy())

plt.imshow(cv2.cvtColor(adjast_contrast_brightness(blurred_img.copy(), alpha=1.3, beta=20), cv2.COLOR_BGR2RGB))
```
