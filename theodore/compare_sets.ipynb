```python
from tqdm import tqdm
import numpy as np

import common
```

```python
ds1 = common.load_dataset("data/training_25k_1.tfrecord", common.parse_feature)
ds4 = common.load_dataset("data/training_25k_4.tfrecord", common.parse_feature)
ds4_mask = common.load_dataset("data/training_25k_sample_4_mask.tfrecord", common.parse_feature)
ds4_weighted_mask = common.load_dataset("data/training_weighted_25k_4_mask.tfrecord", common.parse_feature)
```

```python
all_filenames = []
for ds  in [ds1, ds4, ds4_mask, ds4_weighted_mask]:
    filenames = []
    for example in tqdm(ds):
        filenames.append((example['image/filename'].numpy()).decode('utf-8'))
    all_filenames.append(filenames)
```
