```python
import numpy as np
import pandas as pd
import os
```

```python
df = pd.read_csv("tmp/data_video_paths_upd_nearest_first.csv")
df.shape
```

```python
df[df["video_path"].notna()].shape
```

```python
df["seq_name"] = df["seq_path"].apply(lambda x: os.path.basename(x))
```

```python
df["seq_name"].unique().shape
```

```python
df[df["video_path"].notna()]["seq_name"].unique().shape
```

```python
df[df["video_path"].notna() & df["is_algo_test"]].shape, df[df["is_algo_test"]].shape
```

```python
df[~df["video_path"].notna() & df["is_algo_test"]].shape
```

```python
df = df[df["video_path"].notna()]
df.shape
```

```python
df["video_path"] = df["video_path"].apply(lambda x: x.replace("//", "/"))
```

```python
df["seq_name"].unique().shape, df.shape
```

```python
res = []
for useg_name in df["seq_name"].unique():
    tmp_df = df[df["seq_name"] == useg_name].sort_values(by="seq_path")
    test_tmp_df = tmp_df[tmp_df["is_algo_test"]]
    if test_tmp_df.shape[0] > 0:
        res.append(test_tmp_df.iloc[:1])
    else:
        res.append(tmp_df.iloc[:1])
```

```python
df_unique = pd.concat(res, axis=0)
df_unique.shape
```
