```python
import json
import os
import glob
import sys

import utils_cam4_revb

%load_ext autoreload
%autoreload 2
```

```python
BASE_PTH = "/home/klymy/data/cam4_revB"

SUBSETS_PTH = "train_val_holdout_subsets_batches_1_3.json"
NEW_DATA_BATCHES = ["4.2_batch4", "4.2_batch5"]
```

```python
with open(os.path.join(BASE_PTH, SUBSETS_PTH), "r") as f:
    subsets = json.load(f)
```

```python
holdout_h_locs = [utils_cam4_revb.get_data(os.path.basename(i)).lower() for i in subsets["holdout"]]
holdout_h_locs = [i for i in holdout_h_locs if i != "_"]
```

```python
batch4_5_paths = []
for data_batch in NEW_DATA_BATCHES:
    batch4_5_paths.extend(glob.glob(os.path.join(BASE_PTH, data_batch, "*.tif")))

batch4_5_h_locs = [utils_cam4_revb.get_data(os.path.basename(i)).lower() for i in batch4_5_paths]
print(len(batch4_5_h_locs), "batch4_5_h_locs")
```

```python
holdout_h_locs = set(holdout_h_locs)
batch4_5_h_locs = set(batch4_5_h_locs)
print(len(holdout_h_locs), "holdout_h_locs")
print(len(batch4_5_h_locs), "batch4_5_h_locs")
```

```python
intercected_locs = holdout_h_locs.intersection(batch4_5_h_locs)
intercected_locs
```

```python
with open(os.path.join(BASE_PTH, "batches1_3_vs_4_5_holdout_intersection.json"), "w") as f:
    json.dump(
        {"h_locs": list(intercected_locs)},
    f)
```
