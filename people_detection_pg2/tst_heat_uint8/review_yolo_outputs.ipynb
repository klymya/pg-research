```python
import numpy as np
import os
import glob
```

```python
FP32_BASE_PTH = "head_uint8_saved_model"
INT8_BASE_PTH = "head_uint8_quant_input_norm_saved_model"

outouts_fp32_pths = sorted(glob.glob(os.path.join(FP32_BASE_PTH, "*.npy")))
outouts_int8_pths = sorted(glob.glob(os.path.join(INT8_BASE_PTH, "*.npy")))

outouts_fp32 = [np.load(pth) for pth in outouts_fp32_pths]
outouts_int8 = [np.load(pth) for pth in outouts_int8_pths]

len(outouts_fp32), len(outouts_int8)
```

```python
outouts_fp32[0].shape, outouts_int8[0].shape
```

```python
for idx, i in enumerate(range(len(outouts_fp32))):
    if not np.array_equal(outouts_fp32[i], outouts_int8[i]):
        print(f"Outputs differ at index {i}")
        print(
            f"FP32 mean: {outouts_fp32[i].mean():>8.03f}. std: {outouts_fp32[i].std():>8.03f}. "
            f"min: {outouts_fp32[i].min():>8.03f}. max: {outouts_fp32[i].max():>8.03f}"
        )
        print(
            f"INT8 mean: {outouts_int8[i].mean():>8.03f}. std: {outouts_int8[i].std():>8.03f}. "
            f"min: {outouts_int8[i].min():>8.03f}. max: {outouts_int8[i].max():>8.03f}\n"
        )
```
