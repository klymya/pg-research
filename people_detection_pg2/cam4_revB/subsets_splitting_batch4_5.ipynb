```python
import os
import glob
import json

import numpy as np

import utils_cam4_revb

%load_ext autoreload
%autoreload 2
```

```python
ROOT_PTH = "/home/klymy/data/cam4_revB"

NEW_DATA_BATCHES = ["4.2_batch12"]

INTERSECTION_NAME = "batches1_3_vs_12_holdout_intersection.json"
SPLITTING_NAME = "train_val_holdout_subsets_batches_12.json"

VAL_PERCENT = 0.1
```

```python
img_paths = []
for data_batch in NEW_DATA_BATCHES:
    img_paths.extend(glob.glob(os.path.join(ROOT_PTH, data_batch, "*.tif")))
N = len(img_paths)
N
```

```python
with open(os.path.join(ROOT_PTH, INTERSECTION_NAME), "r") as f:
    holdout_names = json.load(f)["h_locs"]

holdout_names
```

```python
holdout_pths = [i for i in img_paths if utils_cam4_revb.is_in_subset(i, holdout_names)]
working_pths = [i for i in img_paths if i not in holdout_pths]
len(holdout_pths), len(working_pths)
```

```python
np.random.seed(42)
np.random.shuffle(working_pths)
val_pths = working_pths[:int(len(working_pths) * VAL_PERCENT)]
training_pths = working_pths[int(len(working_pths) * VAL_PERCENT):]
len(training_pths), len(val_pths), len(holdout_pths), len(training_pths) + len(val_pths) + len(holdout_pths)
```

```python
with open(os.path.join(ROOT_PTH, SPLITTING_NAME), "w") as f:
    json.dump({
        "train": training_pths,
        "val": val_pths,
        "holdout": holdout_pths,
    }, f)
```
