```python
import json
import os
import glob
import sys

import utils_cam4_revb

%load_ext autoreload
%autoreload 2
```

```python
BASE_PTH = "/home/klymy/data/cam4_revB"

SUBSETS_PTH = "train_val_holdout_subsets_batches_1_3.json"
NEW_DATA_BATCHES = ["4.2_batch8", "4.2_batch9", "4.2_batch10"]

INTERSECTIONS_PTH = "batches1_3_vs_8_10_holdout_intersection.json"

# FILTERED_NAMES = [
#     "4.2_H_cam_240_empty_",
#     "4.2_H_cam_239_MenivimEnt_",
#     "4.2_H_cam_79_menivimOS_"
# ]
```

```python
with open(os.path.join(BASE_PTH, SUBSETS_PTH), "r") as f:
    subsets = json.load(f)
```

```python
holdout_h_locs = [utils_cam4_revb.get_data(os.path.basename(i)).lower() for i in subsets["holdout"]]
holdout_h_locs = [i for i in holdout_h_locs if i != "_"]
```

```python
new_batch_h_locs = []
for data_batch in NEW_DATA_BATCHES:
    new_batch_h_locs.extend(glob.glob(os.path.join(BASE_PTH, data_batch, "*.tif")))

new_batch_h_locs = [utils_cam4_revb.get_data(os.path.basename(i)).lower() for i in new_batch_h_locs]
print(len(new_batch_h_locs), "new_batch_h_locs")
```

```python
holdout_h_locs = set(holdout_h_locs)
new_batch_h_locs = set(new_batch_h_locs)
print(len(holdout_h_locs), "holdout_h_locs")
print(len(new_batch_h_locs), "new_batch_h_locs")
```

```python
intercected_locs = holdout_h_locs.intersection(new_batch_h_locs)
intercected_locs
```

```python
with open(os.path.join(BASE_PTH, INTERSECTIONS_PTH), "w") as f:
    json.dump(
        {"h_locs": list(intercected_locs)},
    f)
```
