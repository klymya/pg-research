```python
ROOT_PTH = "/home/klymy/data/cam4_revB"

NEW_DATA_BATCHES = ["4.2_batch_H270_16-44-09"]
SPLITTING_NAME = "train_val_holdout_subsets_batches_H270_16-44-09.json"
```

```python
img_paths = []
for data_batch in NEW_DATA_BATCHES:
    jpg_imgs = glob.glob(os.path.join(ROOT_PTH, data_batch, "*.jpeg"))
    png_imgs = glob.glob(os.path.join(ROOT_PTH, data_batch, "*.png"))
    tif_imgs = glob.glob(os.path.join(ROOT_PTH, data_batch, "*.tif"))
    img_paths.extend(jpg_imgs + png_imgs + tif_imgs)
N = len(img_paths)
N
```

```python
with open(os.path.join(ROOT_PTH, SPLITTING_NAME), "w") as f:
    json.dump({
        "train": [],
        "val": [],
        "holdout": img_paths,
    }, f)
```
