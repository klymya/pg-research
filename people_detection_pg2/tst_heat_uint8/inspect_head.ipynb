```python
import matplotlib.pyplot as plt
import cv2
import numpy as np
import tensorflow as tf
print("TensorFlow version: ", tf.__version__)
```

```python
def load_model(model_path):
    interpreter = tf.lite.Interpreter(model_path=model_path)
    interpreter.allocate_tensors()
    return interpreter
```

```python
head_fp32 = load_model("head_uint8_saved_model/head_float32.tflite")
head_uint8 = load_model("head_uint8_saved_model/head_full_integer_quant.tflite")
```

```python
for i, j in zip(head_fp32.get_tensor_details(), head_uint8.get_tensor_details()):
    print(i)
    print(j)
    print("====================\n")
```
