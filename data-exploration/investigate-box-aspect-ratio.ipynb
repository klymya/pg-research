```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv("box_sizes.csv")
df.shape
```

```python
df = df[(df["width"] > 0) & (df["height"] > 0)]
```

```python
df.head()
```

```python
df["box_aspect_ratio"] = df["height"] / (df["width"] + 1e-6)
```

```python
sns.displot(data=df, x="box_aspect_ratio", kde=True, bins=15)
```

```python
sns.boxplot(data=df, x="box_aspect_ratio")
```
