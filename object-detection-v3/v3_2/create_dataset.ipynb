```python
# only two classes
dataset_name = "table_chair_table_full_v2"

df = pd.read_csv(os.path.join(BASE_PATH, "annotation_merged_furniture_table_full.csv"))
df_imgs = pd.read_csv(os.path.join(BASE_PATH, "images_grouped_th_0.65_holdout.csv"))

popular_labels = ["chair", "table"]
df = df[df["label"].isin(popular_labels)]

print(df["label"].value_counts())

objects, label_mapping = yolov8_dataset_util.dataframes2objects(df, df_imgs, labels=popular_labels, img_folder_prefix="images")
yolov8_dataset_util.create_yolo_dataset(
    objects, label_mapping, os.path.join(BASE_PATH, "annotation"), 
    os.path.join(os.path.dirname(os.getcwd()), "datasets", "v3_2", dataset_name), 
    os.path.join(BASE_PATH, "train_test_split_v3_compability.json"),
    kpt_shape=[1, 2]
)

yolov8_dataset_util.create_yolo_dataset(
    objects, label_mapping, os.path.join(BASE_PATH, "annotation"), 
    os.path.join(os.path.dirname(os.getcwd()), "datasets", "v3_2", f"{dataset_name}_holdout"), 
    os.path.join(BASE_PATH, "holdout_split.json"),
    kpt_shape=[1, 2]
)
```
