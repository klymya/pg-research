```python
import os
import shutil

from ultralytics import YOLO
import matplotlib.pyplot as plt
import cv2
import tqdm

%matplotlib inline
```

```python
model = YOLO("results/results_baseline_pretrain_theodore/weights/best.pt")
model_occupied = YOLO("results/no_occupied_office_chairs/weights/best.pt")
```

```python
iou = 0.5
conf = 0.25
```

```python
baseline_dataset_pth = "experiments/baseline/images/val"
occupied_dataset_pth = "experiments/no_occupied_office_chairs/images/val"
results_pth0 = "viz_baseline"
results_pth1 = "viz_occupied"

shutil.rmtree(f"{results_pth0}_baseline", ignore_errors=True)
os.makedirs(results_pth0)
shutil.rmtree(f"{results_pth1}_baseline", ignore_errors=True)
os.makedirs(results_pth1)

for img_pth in tqdm.tqdm(os.listdir(baseline_dataset_pth), desc="baseline"):
    img_pth = os.path.join(baseline_dataset_pth, img_pth)

    plt.figure(figsize=(20, 7))
    plt.suptitle(os.path.basename(img_pth), fontsize=15)

    results = model(img_pth, verbose=False, iou=iou, conf=conf)
    plt.subplot(1, 2, 1)
    # Visualize the detection
    img = results[0].plot()  # This plots the detections on the image
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title("baseline")

    results = model_occupied(img_pth, verbose=False, iou=iou, conf=conf)
    plt.subplot(1, 2, 2)
    # Visualize the detection
    img = results[0].plot()  # This plots the detections on the image
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title("baseline+occupied")
    plt.savefig(os.path.join(results_pth1, os.path.basename(img_pth).replace(".bmp", ".jpg")))
    plt.close()

for img_pth in tqdm.tqdm(os.listdir(occupied_dataset_pth), desc="occupied"):
    img_pth = os.path.join(occupied_dataset_pth, img_pth)

    plt.figure(figsize=(20, 5))
    plt.suptitle(os.path.basename(img_pth), fontsize=15)

    results = model(img_pth, verbose=False, iou=iou, conf=conf)
    plt.subplot(1, 2, 1)
    # Visualize the detection
    img = results[0].plot()  # This plots the detections on the image
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title("baseline")

    results = model_occupied(img_pth, verbose=False, iou=iou, conf=conf)
    plt.subplot(1, 2, 2)
    # Visualize the detection
    img = results[0].plot()  # This plots the detections on the image
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title("baseline+occupied")
    plt.savefig(os.path.join(results_pth0, os.path.basename(img_pth).replace(".bmp", ".jpg")))
    plt.close()

# plt.show()
```
