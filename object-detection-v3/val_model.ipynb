```python
import os
import pickle

import numpy as np
import pandas as pd
from ultralytics import YOLO
import tqdm
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
base_pth = "results"
model_name = "v2_batch6_merged_furniture"

thres = 0.4
iou_thres = 0.5
```

```python
model = YOLO(os.path.join(base_pth, model_name, "weights/best.pt"))
metrics = model.val(
    data="datasets/v2_batch6/merged_furniture/project.yaml",
    task="pose",
    iou=iou_thres, conf=thres, device=[1], batch=30, verbose=True
)

print("F1", metrics.box.f1, metrics.pose.f1)
print("avg F1", metrics.box.f1.mean(), metrics.pose.f1.mean())
```

```python
model = YOLO(os.path.join(base_pth, model_name, "weights/best.pt"))
metrics = model.val(
    data="datasets/v2_batch6/merged_furniture_gray/project.yaml",
    task="pose",
    iou=iou_thres, conf=thres, device=[1], batch=30, verbose=True
)

print("F1", metrics.box.f1, metrics.pose.f1)
print("avg F1", metrics.box.f1.mean(), metrics.pose.f1.mean())
```
